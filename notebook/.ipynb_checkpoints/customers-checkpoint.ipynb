```python
import pandas as pd
from ydata_profiling import ProfileReport

# Load raw customers table
df = pd.read_csv(r"D:\business-decision-intelligence-system\data\raw\customers.csv")

# Generate profiling report
profile = ProfileReport(
    df,
    title="Customers – Raw Profiling",
    explorative=True
)

# Save report as HTML
profile.to_file("customers_profiling.html")

```

```python
import pandas as pd
from ydata_profiling import ProfileReport
from pathlib import Path

RAW_DIR = Path(r"D:\business-decision-intelligence-system\data\raw")
REPORT_DIR = Path(r"D:\business-decision-intelligence-system\profiling_reports")

REPORT_DIR.mkdir(exist_ok=True)

for file in RAW_DIR.glob("*.csv"):
    table_name = file.stem
    print(f"Profiling {table_name}...")

    df = pd.read_csv(file)

    profile = ProfileReport(
        df,
        title=f"{table_name} – Raw Data Profiling",
        explorative=True
    )

    profile.to_file(REPORT_DIR / f"{table_name}_profiling.html")
```
